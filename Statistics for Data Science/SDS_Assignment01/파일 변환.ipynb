```python
import os

path = './Data'

print(f"'{path}' 폴더 안의 실제 파일 목록:")

# os.listdir()는 해당 폴더의 모든 파일과 폴더 이름을 리스트로 반환합니다.
file_list_in_os = os.listdir(path)

for filename in file_list_in_os:
    # repr() 함수는 문자열을 표현할 때 숨겨진 문자(공백, \n 등)를 그대로 보여줍니다.
    print(repr(filename))
```

```python
import pandas as pd
import glob
import os

# 1. 변환할 파일들이 있는 폴더 경로를 지정합니다.
#    (실제 파일이 있는 경로로 꼭 수정해주세요!)
path = './data' 

# 2. 해당 폴더에서 '.txt'로 끝나는 모든 파일의 경로를 리스트로 가져옵니다.
txt_file_list = glob.glob(os.path.join(path, '*.txt'))

# --- 변환 작업 시작 ---
print(f"총 {len(txt_file_list)}개의 TXT 파일을 CSV로 변환합니다.")

# 3. 파일 리스트를 순회하며 하나씩 변환 작업을 수행합니다.
for file_path in txt_file_list:
    try:
        # 3-1. TXT 파일을 읽어옵니다. (탭으로 구분, cp949 인코딩)
        df = pd.read_csv(file_path, sep=',', encoding='cp949')
        
        # 3-2. 저장할 CSV 파일의 새로운 경로를 생성합니다.
        #       (예: './data/file.txt' -> './data/file.csv')
        base_name = os.path.splitext(os.path.basename(file_path))[0]
        csv_file_path = os.path.join(path, f"{base_name}.csv")
        
        # 3-3. DataFrame을 CSV 파일로 저장합니다.
        #       index=False: 불필요한 인덱스 열이 생기는 것을 방지합니다.
        #       encoding='utf-8-sig': Excel에서 한글이 깨지지 않게 합니다.
        df.to_csv(csv_file_path, index=False, encoding='cp949')
        
        print(f"  [성공] {os.path.basename(file_path)} -> {os.path.basename(csv_file_path)}")

    except Exception as e:
        # 파일 하나에서 오류가 발생하더라도 전체 스크립트가 멈추지 않도록 처리
        print(f"  [실패] {os.path.basename(file_path)} 파일 변환 중 오류 발생: {e}")

print("--- 모든 작업이 완료되었습니다. ---")
```
